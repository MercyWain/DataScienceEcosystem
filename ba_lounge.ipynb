```python
grouped = df.groupby(['TIME_OF_DAY', 'HAUL', 'ARRIVAL_REGION'])[['TIER1_ELIGIBLE_PAX', 'TIER2_ELIGIBLE_PAX', 'TIER3_ELIGIBLE_PAX']].sum()
grouped
```

```python
grouped_pax = df.groupby(['TIME_OF_DAY', 'HAUL', 'ARRIVAL_REGION'])[['FIRST_CLASS_SEATS', 'BUSINESS_CLASS_SEATS', 'ECONOMY_SEATS']].sum()
grouped_pax['TOTAL_PAX'] = grouped_pax['FIRST_CLASS_SEATS'] + grouped_pax['BUSINESS_CLASS_SEATS'] + grouped_pax['ECONOMY_SEATS']
grouped_pax
```

```python
lookup_table = grouped.copy()
lookup_table['TOTAL_PAX'] = grouped_pax['TOTAL_PAX']

lookup_table['TIER1_PCT'] = lookup_table['TIER1_ELIGIBLE_PAX'] / lookup_table['TOTAL_PAX'] * 100
lookup_table['TIER2_PCT'] = lookup_table['TIER2_ELIGIBLE_PAX'] / lookup_table['TOTAL_PAX'] * 100
lookup_table['TIER3_PCT'] = lookup_table['TIER3_ELIGIBLE_PAX'] / lookup_table['TOTAL_PAX'] * 100

lookup_table
```
